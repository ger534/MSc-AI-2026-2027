# Task 1

The code provided below is for a <b>recursive function</b> that computes the edit distance between two strings <i>x</i> and <i>y</i>. Modify this code such that:

(a) The function measures a <b>weighted edit distance</b>, where

- insertion cost = 2
- deletion cost = 2
- substitution cost = 2, if a letter is substituted with another letter <u>or</u> a digit is substituted with another digit
- substitution cost = 4, if a letter is substituted with a digit or vice versa

(b) use the <u>time.time()</u> method to measure the execution time.

<b>Note 1:</b> If you are not familiar with recursive functions, you must first watch the short optional video in the lesson, which reviews this topic.

<b>Note 2:</b> More advanced coders can consider adding <u>tracing</u> capabilities to their recursive function. This optional task carries no marks but could be an insightful way to understand recursion.

<b>Expected Output</b> - demonstrate the functionality of your function by trying the two test cases shown below:

```
Edit Distance between "intention" and "execution": ??????
--- Executed in: 2.647733449935913 seconds ---

Edit Distance between "ABC" and "456": ??????
--- Executed in: 0.00036716461181640625 seconds ---
```

````python
#TASK 1
import time

#How many insertions, deletions, or substitutions does it take to turn x into y?
def edDistRecursive(x, y):
    # If either x or y is empty, then one or more (depending on their length) INSERTIONS or DELETIONS are needed to transform x to y.
    if len(x) == 0:
        # the original code set the cost for each insertion at 1, the modification requires to multiple this by 2
        #print(f'\t Converting "{x}" TO "{y}" requires {len(y)} INSERTIONS, therefore it costs {len(y)*2}')
        return len(y*2)
    if len(y) == 0:
        # the original code set the cost for each insertion at 1, the modification requires to multiple this by 2
        #print(f'\t Converting "{x}" TO "{y}" requires {len(x)} DELETIONS, therefore it costs {len(x)*2}')
        return len(x*2)

    delta = 0

    # When neither x nor y is empty, we must compute the cost of each operation and find the least costly one.
    if x[-1] != y[-1]:
        # when x and y are the same type (digit to digit or letter to letter), the cost is 2
        if str.isdigit(x) == str.isdigit(y) or str.isalpha(x) == str.isalpha(y):
            delta = 2
        else:
            # when the substitution swaps a number for a letter or viceversa, the cost is 4
            delta = 4

    diagonal_or_substitution_cost = edDistRecursive(x[:-1], y[:-1]) + delta  #what's the cost of SUBSTITUTING the last character of x with the last character of y
    vertical_or_deletion_cost     = edDistRecursive(x[:-1], y)      + 2      #what's the cost of DELETING the last character of x
    horizontal_or_insertion_cost  = edDistRecursive(x, y[:-1])      + 2      #what's the cost of INSERTING the last character of x into y

    # What's the least costly operation?
    minValue = min(diagonal_or_substitution_cost, vertical_or_deletion_cost, horizontal_or_insertion_cost)
    return minValue


# Test cases:
start_time = time.time()
test_result_1  = edDistRecursive("intention", "execution")
end_time = time.time()
print(f"Edit Distance between \"intention\" and \"execution\": {test_result_1}")
#expecting result = del (2), sub letter to letter (2), sub letter to letter (2), ins (2) sub letter to letter (2) = 10
print(f"--- Executed in: {end_time - start_time} seconds ---")
start_time = time.time()
test_result_2  = edDistRecursive("ABC", "456")
end_time = time.time()
print(f"Edit Distance between \"ABC\" and \"456\": {test_result_2}")
print(f"--- Executed in: {end_time - start_time} seconds ---")
````
