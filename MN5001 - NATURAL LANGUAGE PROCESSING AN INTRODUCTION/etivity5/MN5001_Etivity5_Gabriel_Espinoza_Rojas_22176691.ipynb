Sentiment analysis is the interpretation and classification of emotions within text. Write a <b>simple rule-based sentiment analyser</b>, which takes a string as input and returns its sentiment (positive, negative, neutral).

How a rule-based system works:

1. Define two lists of polarised words (e.g. negative words such as <i>bad</i>, <i>worst</i>, <i>ugly</i>, and positive words such as <i>good</i>, <i>best</i>, <i>beautiful</i>).

2. Count the number of positive and negative words that appear in a given text.
3. If positive appearances > negative appearances → positive sentiment, and vice versa. If equal → neutral.

````python
#TASK 1 - load the lexicons
import pandas as pd

!wget http://ptrckprry.com/course/ssd/data/positive-words.txt   #Opinion Lexicon: Positive Words; source: http://www.cs.uic.edu/~liub/FBS/sentiment-analysis.html
!wget http://ptrckprry.com/course/ssd/data/negative-words.txt   #Opinion Lexicon: Negative Words; source: http://www.cs.uic.edu/~liub/FBS/sentiment-analysis.html

filePath1 = "/content/positive-words.txt"
filePath2 = "/content/negative-words.txt"

#Latin-1 (ISO-8859-1) character set. Use chardet lib if needed.
#skiprows drops the licence header at the top of each file.
posWords_df = pd.read_csv(filePath1, sep='\t', header=None, names=['words'], encoding='latin-1', skiprows=[i for i in range(0,34)])
negWords_df = pd.read_csv(filePath2, sep='\t', header=None, names=['words'], encoding='latin-1', skiprows=[i for i in range(0,34)])

print(f'Number of posWords: {len(posWords_df)}   Number of negWords: {negWords_df.size}')
display(posWords_df.head(10))
display(negWords_df.head(10))
````

````python
#TASK 1 - the analyser
def sentimentAnalyzer(string):
    """
    Rule-based sentiment: count polarised words, compare the two counts.

      3. TODO: count how many tokens appear in posWords_df and in negWords_df.
               Sets are far faster than dataframe lookups here:
                   pos_set = set(posWords_df['words'])

      5. TODO: CONFIDENCE SCORE (one line). Think about what the score should do
               at the extremes: all-positive -> 1.0, balanced -> 0.0, no polarised
               words at all -> must not divide by zero.
    """
    
    print(f"string= {string}\n")

    confidence = 0

    # normalising the text (lexicons are lower case) performing the tokenisation
    token_list = string.lower().split(" ")

    negative_appearances = 0
    positive_appearances = 0

    for token in token_list:
      # Count the number of positive and negative words that appear in a given text.      
      print(token)
      if token in negWords_df.values:
        negative_appearances = negative_appearances + 1
      if token in posWords_df.values:
        positive_appearances = positive_appearances + 1

    # method for generating confidence scores (can be achieved in one line)
    confidence = 100 * ((negative_appearances + positive_appearances)+1//len(token_list))
    # Think about what the score should do
    # at the extremes: all-positive -> 1.0, balanced -> 0.0, no polarised
    # words at all -> must not divide by zero.


    # If equal → neutral
    # using the math library to compare floats for almost-equality in Python
    #if math.isclose(negative_appearances, positive_appearances): # is this overkill?
    if negative_appearances == positive_appearances:  
      print(f"neutral sentiment 😶")
    else:
      # If positive appearances > negative appearances → positive sentiment 
      if positive_appearances > negative_appearances:
        print(f"positive sentiment 🥹")
      else:
        # If and vice versa
        print(f"negative sentiment 😞")
        # emojis taken from: https://www.webfx.com/tools/emoji-cheat-sheet/


    print(f"confidence: {confidence}")
    print(f"negative_appearances: {negative_appearances}")
    print(f"positive_appearances: {positive_appearances}")
    
    print("--------------------------------")

sentimentAnalyzer("NLP is cool")
sentimentAnalyzer("NLP is cool and useful")
sentimentAnalyzer("NLP is hard")
sentimentAnalyzer("NLP is hard and useless")
sentimentAnalyzer("NLP stands for Natural Language Processing")
````

## Task 2 [7 marks]

Write a <code>naiveBayesClassifier</code> function which takes the sample training and testing documents below as input, and uses the <b>Naive Bayes</b> algorithm to classify the test documents.

| | Doc | Words | Class |
|---|---|---|---|
| Training | 1 | London is the Capital of GB | GB |
| | 2 | Oxford is a city in GB | GB |
| | 3 | Dublin is the capital of Ireland | IE |
| | 4 | Limerick is a city in Ireland | IE |
| Test | 5 | University of Limerick | ? |
| | 6 | University College Dublin | ? |
| | 7 | Imperial College London | ? |
| | 8 | University of Oxford | ? |
| | 9 | Ireland &amp; GB | ? |

<b>Hint 1:</b> Python dictionaries are suitable data structures for storing BoWs (Bag of Words).

<b>Expected Output:</b>

```
megaDocGB= london...
megaDocIE= dublin...

probGB= ?   probIE= ?

GB_BOW= {'london': 1, ...}
IE_BOW= {'dublin': 1, ...}

V = {'london': 1, ...}
|V| = ?

------------------------------------------------
Test document=('University of Limerick', '?')

word= "..."    wordConditionalProbGB= ?    wordConditionalProbIE= ?
word= "..."    wordConditionalProbGB= ?    wordConditionalProbIE= ?
word= "..."    wordConditionalProbGB= ?    wordConditionalProbIE= ?

docProbGB= ?   docProbIE= ?
```

````python
#TASK 2
import math
from collections import Counter


def naiveBayesClassifier(trainingSet, testSet):
    """
    Multinomial Naive Bayes with add-one (Laplace) smoothing.
    """

    # first the mega-documents Dan talks about, simply just concatenate all documents for each class into just one
    megaDocGB = ""
    megaDocIE = ""
    pseudo_V = ""
    # to calculate priors
    GB_total_count = 0
    IE_total_count = 0

    for pair in trainingSet:
        document = pair[0]
        class_id = pair[1]
        if class_id == "GB":
            GB_total_count = GB_total_count + 1
            megaDocGB = megaDocGB + document + " "
        else:
            # if an input had any other class this would be wrong (other than IE)
            IE_total_count = IE_total_count + 1
            megaDocIE = megaDocIE + document + " "

        # incomplete V, needs to drop duplicates later on
        pseudo_V = pseudo_V + document + " "

    # normalising to lower case (is this the right term?) 
    # and drop the last space concatenated, not needed
    megaDocGB = megaDocGB[:-1].lower()
    megaDocIE = megaDocIE[:-1].lower()
    pseudo_V = pseudo_V[:-1].lower()


    print(f"megaDocGB= {megaDocGB}")
    print(f"megaDocIE= {megaDocIE}")


    # How many times does the class GB ocurrs in our training set out of the how many total documents are there in our training set?
    probGB = GB_total_count/len(trainingSet)
    # How many times does the class EI ocurrs in our training set out of the how many total documents are there in our training set?
    probIE = IE_total_count/len(trainingSet)

    # priors (think of each class as a language model, unigrams)
    print(f"probGB= {probGB}   #probIE= {probIE}")

    GB_BOW = {}
    for token in megaDocGB.split(" "):
        if GB_BOW.get(token) is not None:
            GB_BOW[token] = GB_BOW[token] + 1
        else:
            GB_BOW[token] = 1

    IE_BOW = {}
    for token in megaDocIE.split(" "):
        if IE_BOW.get(token) is not None:
            IE_BOW[token] = IE_BOW[token] + 1
        else:
            IE_BOW[token] = 1

    print(f"GB_BOW= {GB_BOW}")
    print(f"IE_BOW= {IE_BOW}")

    V = {}
    # let's use the pseudo V (minus the extra space) to build the V dictionary
    for token in pseudo_V.split(" "):
        if V.get(token) is not None:
            V[token] = V[token] + 1
        else:
            V[token] = 1

    print(f"V = {V}")
    V_count = len(V)
    print(f"|V| = {V_count}")

    print("------------------------------------------------")
    for test_document in testSet:
        print(f"Test document={test_document}")

        document = test_document[0]

        # do I need this?
        #conditional_probabilities = {}

        GB_conditional_probabilities = 1
        IE_conditional_probabilities = 1

        # every word must be accounted for as a conditional probability (for both classes)
        # test documents also need to be normalised (no CAPS)
        for token in document.lower().split(" "):
            # not really every word tho, Dan said that non words (not present in our lexicon) are complete ignored
            if V.get(token) is not None:
                #P(w_i|c) = \frac{count(w_i, c) + 1}{count(c) + |V|} with c = class and w_i the token
                wordConditionalProbGB = ((GB_BOW.get(token) if GB_BOW.get(token) is not None else 0) + 1)/(sum(GB_BOW.values()) + V_count)
                wordConditionalProbIE = ((IE_BOW.get(token) if IE_BOW.get(token) is not None else 0) + 1)/(sum(IE_BOW.values()) + V_count)
                GB_conditional_probabilities = GB_conditional_probabilities * wordConditionalProbGB
                IE_conditional_probabilities = IE_conditional_probabilities * wordConditionalProbIE
                print(f"word = \"{token}    \", wordConditionalProbGB= {wordConditionalProbGB},    wordConditionalProbIE= {wordConditionalProbIE}")

        # P(GB|test_document) ∝ P(GB) * prod P(w|GB)
        docProbGB = probGB * GB_conditional_probabilities
        # P(IE|test_document) ∝ P(IE) * prod P(w|IE)
        docProbIE = probIE * IE_conditional_probabilities
        print(f"docProbGB= {docProbGB},   docProbIE= {docProbIE}")

        winner = ""
        # https://www.w3schools.com/python/ref_math_isclose.asp
        if math.isclose(docProbGB, docProbIE):
            winner = 'tie'
        elif docProbGB > docProbIE:
            winner = 'GB'
        else:
            winner = 'IE'

        print(f"Inferred class= {winner}")
        print("------------------------------------------------")


    """
        
       
    ------------------------------------------------
    """


trainingSet = [('London is the Capital of GB','GB'),
               ('Oxford is a city in GB','GB'),
               ('Dublin is the capital of Ireland','IE'),
               ('Limerick is a city in Ireland','IE')]

testSet = [('University of Limerick','?'),
           ('University College Dublin','?'),
           ('Imperial College London','?'),
           ('University of Oxford','?'),
           ('Ireland & GB','?')]

naiveBayesClassifier(trainingSet, testSet)
````
