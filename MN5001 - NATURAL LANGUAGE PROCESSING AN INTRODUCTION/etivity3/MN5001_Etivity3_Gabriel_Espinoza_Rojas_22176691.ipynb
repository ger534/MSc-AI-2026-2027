## Task 1.b

Complete the function below to compute the probability of a given sentence using the unigram and bigram counts from 1.a. Do not apply add-one smoothing

```

bigram count for "i hate" = ?
```

````python
def probability(sentence):

    sentence_list = sentence.split(" ")
    
    # part of the expected output
    print(f"sentence= {sentence_list}\n")

    probabilities = []

    for i in range(0, len(sentence_list)-1):
        
        section = sentence_list[i] + " " +sentence_list[i+1]

        # part of the expected output
        print(section+"\n")

        # https://stackoverflow.com/questions/31536835/extract-value-from-single-row-of-pandas-dataframe
        bigram_count = bigrams_df.loc[bigrams_df['bigram'] == section]['count'].values
        unigram_count = unigrams_df.loc[unigrams_df['unigram'] == sentence_list[i]]['count'].values

        # I had to add logic to avoid errors when the words are not found in the txt files
        # https://stackoverflow.com/questions/5086178/how-to-check-if-array-is-not-empty
        bigram_count = bigram_count[0] if bigram_count.size else 0
        unigram_count = unigram_count[0] if unigram_count.size else 0
       
        print(f"bigram count for \"{section}\" = {bigram_count}")
        print(f"unigram count for \"{sentence_list[i]}\" = {unigram_count}")

        # https://youtu.be/UyC0bBiZY-A?t=16 P(w_i | w_i-1) = count(W_i-1, w_i) / count(W_i-1)
        bigram_probability = bigram_count / unigram_count if unigram_count != 0 else 0
        print(f"bigramProbability for \"{section}\" = {bigram_probability}\n")

        probabilities = probabilities + [bigram_probability]


    """
    Example from chapter 3, bigram
    P(<s> i want english food </s>) =   P(i|<s>)*P(want|i)*P(english|want)*P(food|english)*P(</s>|food)
                                    =   0.25*0.33*0.0011*0.5*0.68
                                    =   0.000031
    """
    sentence_probability = 1
    for probability in probabilities:
        sentence_probability = sentence_probability * probability
    print(f"sentenceProbability for \"{sentence}\" {sentence_probability}\n")
    return sentence_probability

#TESTING
probA = probability("i love you")
probB = probability('i hate you')
#probability("i love you")>probability("i hate you") is: [True,False]?
print(f'probability("i love you")>probability("i hate you") is: {probA>probB}\n')
probability("OOV1 OOV2") # Test for OOV (Out Of Vocabulary) N-grams
````

## Task 1.c

Copy the function from 1.b into a new code cell, and modify it such that it applies add-one smoothing. Your new function should also address the problem of underflow when computing a sentence Probability

````python
"""
from the book:
3.6.1 Laplace Smoothing
The simplest way to do smoothing is to add one to all the n-gram counts, before we normalize them into probabilities. 

https://youtu.be/Hr6UWRwAIEQ?list=PLAUFqkTvH_MQ&t=73
"""

import math

def new_probability(sentence):
    """
    uses add one smoothing (also known as Laplace smoothing).
    The smoothing ensures no bigram will have a probability of zero.
    However, due to the mathematical nature of the interval [0, 1], 
    the probabilities of long sentences would underflow and become zero.
    To address the underflow, the function uses log. 
    """

    sentence_list = sentence.split(" ")
    
    # part of the expected output
    print(f"sentence= {sentence_list}\n")

    probabilities = []

    laplace_smoothing = 1
    V = len(unigrams_df)
    
    for i in range(0, len(sentence_list)-1):
        
        section = sentence_list[i] + " " +sentence_list[i+1]
       
        # part of the expected output
        print(section+"\n")

        # https://stackoverflow.com/questions/31536835/extract-value-from-single-row-of-pandas-dataframe
        bigram_count = bigrams_df.loc[bigrams_df['bigram'] == section]['count'].values
        unigram_count = unigrams_df.loc[unigrams_df['unigram'] == sentence_list[i]]['count'].values
        
        # I had to add logic to avoid errors when the words are not found in the txt files
        # https://stackoverflow.com/questions/5086178/how-to-check-if-array-is-not-empty
        bigram_count = bigram_count[0] if bigram_count.size else 0
        unigram_count = unigram_count[0] if unigram_count.size else 0
       
        print(f"bigram count for \"{section}\" = {bigram_count}")
        print(f"unigram count for \"{sentence_list[i]}\" = {unigram_count}")

        # P(w_i | w_i-1) = count(W_i-1, w_i) + 1 / count(W_i-1) + V (https://youtu.be/UyC0bBiZY-A?t=16)
        bigram_probability = (bigram_count + laplace_smoothing) / (unigram_count + V)
        print(f"bigramProbability for \"{section}\" = {bigram_probability}\n")

        probabilities = probabilities + [bigram_probability]

    # addressing underflow using log, instead of multiplying, we sum the log of the probabilities
    sentence_probability = 1
    sentence_log_probability = 0
    for probability in probabilities:
        sentence_probability = sentence_probability * probability
        sentence_log_probability = sentence_log_probability + math.log(probability)
    
    print(f"sentenceProbability for \"{sentence}\" {sentence_probability}")
    print(f"Log sentenceProbability for \"{sentence}\" {sentence_log_probability}")
    print(f"Does {sentence_probability} equals exp({sentence_log_probability})? {math.exp(sentence_log_probability)}\n")

    # even when it is possible to return the sentence probability, I'm still returning the log probability
    # it is the only way to avoid losing meaningful information when the probability is very small (underflow)
    return sentence_log_probability


#TESTING
probA = new_probability("i love you")
probB = new_probability('i hate you')
print(f'probability("i love you")>probability("i hate you") is: {probA>probB}\n')
new_probability("OOV1 OOV2") # Test for OOV (Out Of Vocabulary) N-grams
````

Using the [bigram counts](https://www.norvig.com/ngrams/count_2w.txt) (bigrams_df), write a function that implements the Shannon visualisation method:

[Chapter 3, section 3.4](https://web.stanford.edu/~jurafsky/slp3/3.pdf) gives a detailed description of the Shannon Visualisation method; a screenshot of this section is provided below. If you can follow the logic applied to unigrams, then you should be able to apply the same logic to bigrams.

Note: given the large size of the bigrams data, the Shannon Visualisation sampling can continue for a long time; therefore, for initial testing, you could terminate the sampling early whenever a common word (e.g. “and”) is generated as the next new seed.

Extra Help: 
- The seed word can be any word, but let's stick to the default seed word `<S>`, which indicates the beginning of the sentence. 
- First, you need to retrieve a list of all the bigrams starting with `<S>`, and then <u>randomly select one of them according to its probability</u>. Understanding this last part is key! You have a probability space between 0 and 1, and each of the candidate bigrams covers <u>an interval in this space proportional to its frequency</u>. For example, the candidate bigram “`<S>` and” has a total frequency of 261891475, which translates to a probability of 0.031513. If you consider this bigram to be your first candidate, then you can assign the probability interval 0.0 to 0.031513 to it. There are 8640 bigrams in the dataset which start with `<S>` and you need to compute the probability intervals for all of them. 
- Then generate a random number between 0 and 1 and check to see which interval it falls into; that would be your chosen bigram.
- Pick the second word in the chosen bigram, and that would be your new seed word.
- This process should be repeated until you either reach `</S>` (indicating the end of the sentence) or come across a seed which does not exist in your list of bigrams (i.e. there is no bigram starting with this seed word).


````python
import random

def ShannonVisualization(full_sentence="", seed="<S>"):

    # this is never going to happen because </S> is not part of the e-tivity corpus
    # however, other datasets may have it since it is a common practice to use it as the ending token
    if seed == "</S>":
        return full_sentence
    
    # get all the bigrams starting with the seed word
    bigrams_starting_with_seed_word_df = bigrams_df.loc[bigrams_df['bigram'].str.contains("^" + seed + " ")]
    
    # resetting the index to avoid any confusions (not sure if this is the best way to go or adding a table and appending the probability to each row would be better)
    bigrams_starting_with_seed_word_df = bigrams_starting_with_seed_word_df.reset_index()

    # this will hold the sum of all bigrams' counts with the seed word
    full_count = 0
    for index, row in bigrams_starting_with_seed_word_df.iterrows():
        full_count = full_count + row['count']

    if full_count == 0:
        # a seed which does not exist in your list of bigrams (i.e. there is no bigram starting with this seed word).
        return full_sentence

    probabilities = []

    for index, row in bigrams_starting_with_seed_word_df.iterrows():
        probabilities = probabilities + [row['count']/full_count]

    # this will hold the sum of probabilities
    cumulative_probabilities = 0

     # this is the number between 0 and 1 that will decide the starting bigram
    random_pick = random.random()
    
    # this loop will increment the probability until it is greater than the random number, this is how we find the selected bigram and thefore the next seed
    for index, probability in enumerate(probabilities):

        cumulative_probabilities = cumulative_probabilities + probability
        
        if random_pick < cumulative_probabilities:

            # setting the amount of spacing accordingly to give that waterfall effect
            print( " "*len(full_sentence) + bigrams_starting_with_seed_word_df.iloc[index]['bigram'])
            next_seed = bigrams_starting_with_seed_word_df.iloc[index]['bigram'].split(" ")[1]

            #recursion since it is a recurrent theme in the module
            return ShannonVisualization(full_sentence + next_seed + " ", next_seed)


final_sentence = ShannonVisualization()
print(final_sentence)
````
