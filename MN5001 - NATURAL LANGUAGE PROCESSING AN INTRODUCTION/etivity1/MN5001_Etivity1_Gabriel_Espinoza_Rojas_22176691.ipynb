## Task 1.1

Write a function in Python that takes a string as input and uses regular expressions to:
-  &nbsp;
    - A. check if the given string is a valid Eircode or not
    - B. if the string is a valid Eircode, identify and print out the Eircode's geographical district.

- Each Eircode consists of seven CAPITAL letters and digits.
- The format of the Eircode is a 3-digit alphanumeric “routing key”, followed by a 4-digit alphanumeric "unique identifier".
- The Routing Key will always start with a letter A, C, D, E, F, H, K, N, P, R, T, V, W, X or Y, and will be followed by two numeric digits (0-9) - except for the area of the D6W where the letter W is valid on the 3rd digit.
- The Unique Identifier will comprise a mixture of letters and/or numbers - letters A, C, D, E, F, H, K, N, P, R, T, V, W, X or Y, and numbers 0-9.

- An optional space or hyphen could separate the Routing Key from the Unique Identifier, for example, “V94 T9PX”, “V94-T9PX”


* You can download the full list of routing keys in CSV format from HERE using the !wget command.
* Use the 're' library for regular expressions in Python.
* Regex101 is a handy website for trying and testing your regular expressions.


```python
import re   # import re library for regex
import csv  # import csv library to handle the Eircode CSV file


# use the linux command wget to donwload the CSV file
!wget https://gist.githubusercontent.com/ajoorabchi/eac194a79dd26de8864f9206b7842ff1/raw/8ea1d8d5f74b5b2724e378b43d4df6094990c7db/Eircode%2520Routing%2520Key%2520Boundaries.csv
filePath = "/content/Eircode Routing Key Boundaries.csv" # set the path for the donwloaded CSV file


with open(filePath, 'r') as f:
    reader = csv.reader(f)
    #the map function iterates through the rows in the CSV file and puts them in tuples. The tuples are then added to the KeyRouteList    [(k0,d0), ...,(kx,dx)]
    KeyRouteList = list(map(tuple, reader)) 
#print(KeyRouteList)


def eircodeValidator(eircode):
  """
  Eircode must be seven-character alphanumeric code, 
  The first three characters are the routing key, which is used to identify the general area of the address.  (variable KeyRouteList)
  Then a space or dash 
  The last four characters are the unique identifier, which is used to identify a specific address within that area.
  More info: https://www.eircode.ie/faqs
  """
  print(f"\nEircode: {eircode}")

  # the first character is always a capital letter, followed always by a second capital letter in the exception of (D6W)
  # there might be an optional space or dash, and the last 4 parts are alphanumeric characters.
  # spaces might be before or after the Eircode
  
  match = re.match(r'\s*?([A-Z]{1}[W0-9]{2}[-\s]?[A-Z0-9]{4})\s*?', eircode)
  if match:
    
   #simply using 'routing_key = eircode[0:3]' didn't work for cases with spaces
    routing_key = match.group(1)[0:3]

    destination_string = f"\n{routing_key} is a valid but unassigned routing key"
    

    # check the routing key against KeyRouteList
    for key, destination in KeyRouteList:
      if routing_key == key:
        destination_string = f"\nDestination= {destination}"
        break

    # start the unique identifier from the 4th character if the Eircode has the optional space or dash, otherwise start from the 3rd character
    unique_identifier = match.group(1)[4:8] if len(match.group(1)) == 8 else match.group(1)[3:7]

    print(f"Valid Eircode pattern, Routing Key= {routing_key}, Unique Identifier= {unique_identifier}" + destination_string)
  
  else:
    print("Invalid Eircode pattern")
  

# Your eircodeValidator function should be able to handle the test cases below:
eircodeValidator("D6W-19PX")
eircodeValidator("V94-T9PX")
eircodeValidator("A91 T9PX")
eircodeValidator("          A93T9PX           ")


eircodeValidator("D6Z-T9PX")
eircodeValidator("123-T9PX")


eircodeValidator("V94 -T9PX")
eircodeValidator("V94,T9PX")


eircodeValidator("v94 T9px")


eircodeValidator("V94T9PXV")
eircodeValidator("V94T9P")
```
