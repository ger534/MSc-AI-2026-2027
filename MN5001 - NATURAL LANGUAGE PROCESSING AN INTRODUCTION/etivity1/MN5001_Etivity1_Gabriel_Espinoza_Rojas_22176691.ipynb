## Task 1.1

Write a function in Python that takes a string as input and uses regular expressions to:
-  &nbsp;
    - A. check if the given string is a valid Eircode or not
    - B. if the string is a valid Eircode, identify and print out the Eircode's geographical district.

- Each Eircode consists of seven CAPITAL letters and digits.
- The format of the Eircode is a 3-digit alphanumeric “routing key”, followed by a 4-digit alphanumeric "unique identifier".
- The Routing Key will always start with a letter A, C, D, E, F, H, K, N, P, R, T, V, W, X or Y, and will be followed by two numeric digits (0-9) - except for the area of the D6W where the letter W is valid on the 3rd digit.
- The Unique Identifier will comprise a mixture of letters and/or numbers - letters A, C, D, E, F, H, K, N, P, R, T, V, W, X or Y, and numbers 0-9.

- An optional space or hyphen could separate the Routing Key from the Unique Identifier, for example, “V94 T9PX”, “V94-T9PX”


* You can download the full list of routing keys in CSV format from HERE using the !wget command.
* Use the 're' library for regular expressions in Python.
* Regex101 is a handy website for trying and testing your regular expressions.


```python
import re   # import re library for regex
import csv  # import csv library to handle the Eircode CSV file


# use the linux command wget to donwload the CSV file
!wget https://gist.githubusercontent.com/ajoorabchi/eac194a79dd26de8864f9206b7842ff1/raw/8ea1d8d5f74b5b2724e378b43d4df6094990c7db/Eircode%2520Routing%2520Key%2520Boundaries.csv
filePath = "/content/Eircode Routing Key Boundaries.csv" # set the path for the donwloaded CSV file


with open(filePath, 'r') as f:
    reader = csv.reader(f)
    #the map function iterates through the rows in the CSV file and puts them in tuples. The tuples are then added to the KeyRouteList    [(k0,d0), ...,(kx,dx)]
    KeyRouteList = list(map(tuple, reader))
#print(KeyRouteList)


def eircodeValidator(eircode):
  """
  Eircode must be seven-character alphanumeric code,
  The first three characters are the routing key, which is used to identify the general area of the address.  (variable KeyRouteList)
  Then a space or dash
  The last four characters are the unique identifier, which is used to identify a specific address within that area.
  More info: https://www.eircode.ie/faqs
  """
  print(f"\nEircode: {eircode}")

  # the first character is always a capital letter, followed always by a second capital letter in the exception of (D6W)
  # there might be an optional space or dash, and the last 4 parts are alphanumeric characters.
  # spaces might be before or after the Eircode

  match = re.match(r'\s*?([A-Z]{1}[W0-9]{2}[-\s]?[A-Z0-9]{4})\s*?', eircode)
  if match:

   #simply using 'routing_key = eircode[0:3]' didn't work for cases with spaces
    routing_key = match.group(1)[0:3]

    destination_string = f"\n{routing_key} is a valid but unassigned routing key"


    # check the routing key against KeyRouteList
    for key, destination in KeyRouteList:
      if routing_key == key:
        destination_string = f"\nDestination= {destination}"
        break

    # start the unique identifier from the 4th character if the Eircode has the optional space or dash, otherwise start from the 3rd character
    unique_identifier = match.group(1)[4:8] if len(match.group(1)) == 8 else match.group(1)[3:7]

    print(f"Valid Eircode pattern, Routing Key= {routing_key}, Unique Identifier= {unique_identifier}" + destination_string)

  else:
    print("Invalid Eircode pattern")


# Your eircodeValidator function should be able to handle the test cases below:
eircodeValidator("D6W-19PX")
eircodeValidator("V94-T9PX")
eircodeValidator("A91 T9PX")
eircodeValidator("          A93T9PX           ")


eircodeValidator("D6Z-T9PX")
eircodeValidator("123-T9PX")


eircodeValidator("V94 -T9PX")
eircodeValidator("V94,T9PX")


eircodeValidator("v94 T9px")


eircodeValidator("V94T9PXV")
eircodeValidator("V94T9P")
```

```python
import re   # import re library for regex
!pip install html2text
import html2text # import html2text library to convert and extract the text content of the HTML file


!wget https://www.ul.ie/contact-information # Use the Linux command wget to download the webpage
filePath = "/content/contact-information" # Set the path for the downloaded HTML file
contact_information_file = open(filePath, "r")
contact_information_html = contact_information_file.read()
contact_information_text = html2text.html2text(contact_information_html)
#print(contact_information_text) #uncomment to see the text content of the page


def contactsExtractor(contact_information_text):

  # dictionary to store UL contacts (phone numbers, email addresses and URLs)
  ul_info = {
      "phone_numbers": [],
      "email_addresses": [],
      "urls": []
  }

  # special case for emergency number (as a extension or local number)
  ul_info["phone_numbers"] = re.findall(r'\+353[-]61[-][0-9]{6}|[0-9]{13}|(?:06121)?3333', contact_information_text)

  # using the UL domain should fit the problem perfectly
  ul_info["email_addresses"] = re.findall(r'([a-zA-Z0-9]+@ul\.ie)', contact_information_text)

  # excluding some URLs with specific keywords to avoid false positives (e.g. office365, eufunds, tel:, euemerge, trustedireland)
  ul_info["urls"] = re.findall(r'(?:https?://|www\.)(?!.*\b(?:office365|eufunds|tel:|euemerge|trustedireland)\b)[^\s)\]]+', contact_information_text)

  # using set to remove duplicates and then convert back to list
  ul_info["phone_numbers"] = list(set(ul_info["phone_numbers"]))
  ul_info["email_addresses"] = list(set(ul_info["email_addresses"]))
  ul_info["urls"] = list(set(ul_info["urls"]))

  print(f"{len(ul_info['phone_numbers'])} Phone Numbers found: {ul_info['phone_numbers']}")
  print(f"{len(ul_info['email_addresses'])} Email Addresses found: {ul_info['email_addresses']}")
  print(f"{len(ul_info['urls'])} URLs found: {ul_info['urls']}")

  # performance evaluation (Type I and Type II errors) of the contactsExtractor function
  """
  False positives - Type I - matching strings that should not match
  False negative - Type II - not matching strings that should match
  """
  # manually scanned the text to find the number of phone numbers, email addresses and URLs. The results are:
  ground_truth_phone_numbers = 10 # not counting the fax number as a phone number
  ground_truth_email_addresses = 6 # not counting repeated email addresses
  ground_truth_urls = 18 # I considered certain URLs not UL-related

  if len(ul_info['phone_numbers']) > ground_truth_phone_numbers:
    print(f"Type I error: {len(ul_info['phone_numbers']) - ground_truth_phone_numbers} false positive(s) in phone numbers")
  elif len(ul_info['phone_numbers']) < ground_truth_phone_numbers:
    print(f"Type II error: {ground_truth_phone_numbers - len(ul_info['phone_numbers'])} false negative(s) in phone numbers")
  else:
    print("No Type I or Type II errors in phone numbers")

  if len(ul_info['email_addresses']) > ground_truth_email_addresses:
    print(f"Type I error: {len(ul_info['email_addresses']) - ground_truth_email_addresses} false positive(s) in email addresses")
  elif len(ul_info['email_addresses']) < ground_truth_email_addresses:
    print(f"Type II error: {ground_truth_email_addresses - len(ul_info['email_addresses'])} false negative(s) in email addresses")
  else:
    print("No Type I or Type II errors in email addresses")

  if len(ul_info['urls']) > ground_truth_urls:
    print(f"Type I error: {len(ul_info['urls']) - ground_truth_urls} false positive(s) in URLs")
  elif len(ul_info['urls']) < ground_truth_urls:
    print(f"Type II error: {ground_truth_urls - len(ul_info['urls'])} false negative(s) in URLs")
  else:
    print("No Type I or Type II errors in URLs")

contactsExtractor(contact_information_text)
```

## Task 2.2
Use the <u>tf.keras.preprocessing.text.Tokenizer</u> class to:


1. Tokenize the Shakespeare corpus
2. Print out the total number of  <u>Tokens</u> in the corpus
3. print out the total number of <u>Types</u> in the corpus
4. Print out the top 10 most frequent Types in the corpus along with their ranks and frequencies. For example:


    "the"	 is ranked #1, with a frequency of xxxxx


```python
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer()
tokenizer.fit_on_texts(ShakespeareContent)

for_token_count = tokenizer.word_counts

tokens = 0
for token in for_token_count:
    tokens += for_token_count[token]

# the video where Shakespeare's corpus is described was key to know what the counts were supposed to be
print("Tokens:", tokens)
print("Types:", len(tokenizer.word_index))

# print o the top 10 most common words and their counts
word_counts = tokenizer.word_counts
sorted_word_counts = sorted(word_counts.items(), key=lambda x: x[1], reverse=True)
print("Top 10 most common words and their counts:")
for index, (word, count) in enumerate(sorted_word_counts[:10], start=1):
    print(f"\"{word}\" is ranked #{index}, with a frequency of {count}")
```
