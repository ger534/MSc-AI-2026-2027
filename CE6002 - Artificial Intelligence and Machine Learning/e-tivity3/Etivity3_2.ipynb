````python
def custom_formatter(x):
    if x == 0:
        return "decision boundary"
    else:
        return "margin"

def plot_margin(X, y, clf):
    plt.scatter(X[y==0][:,0], X[y==0][:,1], label="Class 0")
    plt.scatter(X[y==1][:,0], X[y==1][:,1], label="Class 1")


    # plot the decision function
    ax = plt.gca()
    ax.set_aspect(aspect=1)
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()

    # create grid to evaluate model
    xx = np.linspace(xlim[0], xlim[1], 30)
    yy = np.linspace(ylim[0], ylim[1], 30)
    YY, XX = np.meshgrid(yy, xx)
    xy = np.vstack([XX.ravel(), YY.ravel()]).T
    Z = clf.decision_function(xy).reshape(XX.shape)

    # plot decision boundary and margins
    margins = ax.contour(XX, YY, Z, colors='k', levels=[-1, 0, 1], alpha=0.5,
               linestyles=['--', '-', '--'])
    
    ax.clabel(margins, fmt=custom_formatter, inline=True, fontsize=8)
    
    # plot support vectors
    ax.scatter(clf.support_vectors_[:, 0], clf.support_vectors_[:, 1], s=100,
               linewidth=1, facecolors='none', edgecolors='k', label="Support Vectors")
    ax.legend()
````

### 2. Perform classification on this dataset using an SVM with linear kernel choosing the C parameter to obtain a hard margin.

### 3. Perform classification on this dataset using an SVM with linear kernel choosing the C parameter to obtain a soft margin with a maximum of 15 support vectors.
    - In a plot containing the data, plot the SVM margins indicating the support vectors.
    - Extract the numerical representation of each support vector from the trained model.
    - Obtain the performance of your model on training data.
    - Using information provided by your trained model, identify which of the support vectors lie inside the margin (i.e. are soft margin support vectors).

````python
#default 1.0, but we want to use a soft margin SVM, so we set C to a very small value
soft_margin_C = 0.70

linear_support_vector_classifier_soft_margin_task1 = svm.SVC(C=soft_margin_C, kernel='linear', degree=3, gamma='scale', coef0=0.0, shrinking=True, probability=False, tol=0.001, 
                                                  cache_size=200, class_weight=None, verbose=False, max_iter=-1, decision_function_shape='ovr', break_ties=False, random_state=0)


linear_support_vector_classifier_soft_margin_task1.fit(X_Task1, y_Task1)


print(f"Linear kernel with soft margin (C={soft_margin_C})")
plt.figure(figsize=(10,10))
plot_margin(X_Task1, y_Task1, linear_support_vector_classifier_soft_margin_task1)
plt.title("SVM margins and support vectors for Task 1")
plt.xlabel("x_1")
plt.ylabel("x_2")
plt.grid()
plt.show()


support_vectors = linear_support_vector_classifier_soft_margin_task1.support_vectors_
support_vector_indices = linear_support_vector_classifier_soft_margin_task1.support_

# y classes are assigned to values 0 and 1, but for an accurate numerical representation we need [-1 1]
y_signed = y_Task1*2-1
# distance from the decision boundary for each support vector
distances =  linear_support_vector_classifier_soft_margin_task1.decision_function(support_vectors)
slacks = np.maximum(0, 1 - y_signed[support_vector_indices] * distances)  

# In sklearn, dual_coef_ is y_n * alpha_n
alphas = np.abs(linear_support_vector_classifier_soft_margin_task1.dual_coef_[0])

print("Numerical representation of the support vectors:")
for i, support_vector in enumerate(support_vectors):
    # let's take the alpha values at face value
    if 0 < alphas[i] and alphas[i] < soft_margin_C:
        print(f"Support Vector {i+1}: x_1[{support_vector[0]:.4f}] x_2[{support_vector[1]:.4f}], On the margin")
    elif alphas[i] == soft_margin_C and slacks[i] <= 1:
        print(f"Support Vector {i+1}: x_1[{support_vector[0]:.4f}] x_2[{support_vector[1]:.4f}], Inside margin")
    else:
        print(f"Support Vector {i+1}: x_1[{support_vector[0]:.4f}] x_2[{support_vector[1]:.4f}], Misclassified")

    print(f"    slack: {slacks[i]:.3f}, alpha: {alphas[i]}, y (signed): {y_signed[support_vector_indices[i]]}")


print("Performance:")
# check accuracy // E_in
accuracy = np.mean(linear_support_vector_classifier_soft_margin_task1.predict(X_Task1) == y_Task1)       
print(f"Accuracy: {accuracy * 100:.2f}%")
print(f"E_in: {1 - accuracy:.4f}")

print("15 support vectors max?", 15 >= len(linear_support_vector_classifier_soft_margin_task1.support_vectors_))
````

<!--Margin support vectors are introduced as those support vectors lying exactly on the margin. For these support vectors, the corresponding  will be strictly smaller than C. The other type of support vector is the non-margin support vector. These can lie inside the margin on either side of the median or even outside the margin but on the opposite side! For these support vectors, the alpha will always be equal to C.-->

To better explain the impact of the C parameter over the margin around the decision boundary created by SVMs, it is vital to first see the expression that captures what are SVMs after.
Maximasing the margins to make them wider, can be rewritten as minimising $\frac{w^T w}{2}$. And this would be enough for linearly separable data sets, but real-world problems are often not that simple. That's where a second term can be added to the expression and where C can help us to regularise. $ min \frac{w^T w}{2} + C \sum{n=1}^{N}\xi_n$, this second term introduces soft margins, where the 'slack' variable ($\xi$) is the distance between the vector (inside the margin) and the decision boundary. C, just as we saw in previous notebooks, works like lambda ($C=\frac{1}{\lambda}$), when C is very large, essentially, there's no regularisation acting, so the SVM defaults back to $ min \frac{w^T w}{2} \xi_n$, when C is small, we allow the sum of all $\xi s$ to be some number greater than 0, but smaller than C.




    2. How and why does the width of the margin affect the expected generalisation abilities of the model? (100 words max)

Narrow margins 'listen' more to the data, which we know as overfitting. Wider margins might not have an $E_in$ as good as a narrow margin, but they tend to generalise better (lower $E_out$), they also give me model the advantage to work in <i>almost</i> linearly separable data sets, without the need of kernels/transformations. 

    3. For the soft-margin model, provide the coordinates of the support vectors that do not violate the margin. (100 words max)

Using a soft-marning, we allow certain points to be inside the margin or even misclassified, in both cases, these could end up as support vectors that violate the margins. Support vectors that do not violate the margins sit exactly at the margin. Considering Lagrange multipliers and the optimisation of the dual form, we know that the support vectors have an $\alpha$ bigger or equal to zero, and in particular, for support vectors at the margin $\alpha < C$, in our case $\alpha < 0.7$ (adds no 'slack' $\xi$ to the hinge loss function).

````python
def plot_margin(X, y, clf, ax=None):
    ax.scatter(X[y==0][:,0], X[y==0][:,1], s=30, label="Class 0")
    ax.scatter(X[y==1][:,0], X[y==1][:,1], s=30, label="Class 1")


    # plot the decision function
    ax = plt.gca() if ax is None else ax
    ax.set_aspect(aspect=1)
    xlim = (X[:,0].min()-0.5, X[:,0].max()+0.5)
    ylim = (X[:,1].min()-0.5, X[:,1].max()+0.5)

    # create grid to evaluate model
    xx = np.linspace(xlim[0], xlim[1], 30)
    yy = np.linspace(ylim[0], ylim[1], 30)
    YY, XX = np.meshgrid(yy, xx)
    xy = np.vstack([XX.ravel(), YY.ravel()]).T
    Z = clf.decision_function(xy).reshape(XX.shape)

    # plot decision boundary and margins
    ax.contour(XX, YY, Z, colors='k', levels=[-1, 0, 1], alpha=0.5,
               linestyles=['--', '-', '--'])
    # plot support vectors
    ax.scatter(clf.support_vectors_[:, 0], clf.support_vectors_[:, 1], s=100,
               linewidth=1, facecolors='none', edgecolors='k', label="Support Vectors")
    ax.legend()
````

## Task 3 code
This wrapper is required in below BayesSearchCV for a neural network to combine the two layer arguments (or more if you wish to experiment with more than two layers) into the single hidden_layer_sizes hyper-parameter.

````python
class MLPWrapper(BaseEstimator, ClassifierMixin):
    
    def __init__(self, 
                 layer1=100, 
                 layer2=100, 
                 alpha=0.0001, 
                 activation='relu', 
                 learning_rate_init=1e-3,
                 early_stopping = False,
                 n_iter_no_change = 10,
                solver = 'adam'):
        self.layer1 = layer1
        self.layer2 = layer2
        self.alpha = alpha
        self.activation = activation
        self.learning_rate_init = learning_rate_init
        self.early_stopping = early_stopping
        self.n_iter_no_change = n_iter_no_change
        self.solver = solver

    def fit(self, X, y):
        model = MLPClassifier(
            hidden_layer_sizes=[self.layer1, self.layer2],
            alpha=self.alpha,
            max_iter = 1000,
            activation = self.activation,
            learning_rate_init = self.learning_rate_init,
            early_stopping = self.early_stopping,
            n_iter_no_change = self.n_iter_no_change,
            solver = self.solver,
            random_state=10
        )
        model.fit(X, y)
        self.model = model
        return self

    def predict(self, X):
        return self.model.predict(X)

    def score(self, X, y):
        return self.model.score(X, y)
    
    def get_model():
        return self.model
````

### 3. Train a SVM model and optimise appropriate hyper parameters
    - Use 2 searches: try both Grid-based searches and Bayesian searches.
````python
#Search 1:

bayes_search_for_SVM = BayesSearchCV(
    svm.SVC(random_state=0),
    search_spaces=[
    {   "kernel": ["rbf"],
        "C": Real(1e-3, 1e+3, prior='log-uniform'), 
        "gamma": Real(1e-6, 1e+1, prior='log-uniform'), #Categorical(['scale', 'auto']),
        #'tol': Real(1e-4, 1e-2, prior='log-uniform'),
        # 'max_iter': Integer(1,100), # tones of warning on console
    },
    {   "kernel": ["poly"],
        "C": Real(1e-3, 1e+3, prior='log-uniform'),
        "degree": Integer(1, 8),
        "gamma": Real(1e-6, 1e+1, prior='log-uniform'),
        #"coef0": Real(1e-3, 1e+3, prior='log-uniform')
    },
    {   "kernel": ["sigmoid"],
        "C": Real(1e-3, 1e+3, prior='log-uniform'),
        "gamma": Real(1e-6, 1e+1, prior='log-uniform'),
        #"coef0": Real(1e-3, 1e+3, prior='log-uniform')
    }],
    n_iter=50,
    cv=StratifiedKFold(n_splits=10, shuffle=True, random_state=0),
    scoring="accuracy",
    random_state=0, 
    verbose=0,
    n_jobs=-1
)


start = time.time()

# executes grid search optimization
_ = bayes_search_for_SVM.fit(X_Task3_training, y_Task3_training)

end = time.time()
print(f"time taken searching: {end - start:.2f} seconds")

print("Score on training set:", bayes_search_for_SVM.score(X_Task3_training, y_Task3_training))
print("Score on test set:", bayes_search_for_SVM.score(X_Task3_test, y_Task3_test))
print("Number of support vectors: ", len(bayes_search_for_SVM.best_estimator_.support_vectors_))

bayes_search_for_SVM.best_params_
````
