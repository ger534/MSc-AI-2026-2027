```python
import numpy as np
#from numpy import count_nonzero
from sklearn.feature_extraction.text import CountVectorizer

def my_search_function(corpus, terms):
    """
    # takes a corpus
    # a unigram or bigramm
    # applies AND operations
    # prints out the list and content of matching docs which contain both search terms
    """

    # about N-grams
    """
    ngram_rangetuple (min_n, max_n), default=(1, 1)
    The lower and upper boundary of the range of n-values for different word n-grams or char n-grams to be extracted. 
    All values of n such such that min_n <= n <= max_n will be used. 
    For example an ngram_range of (1, 1) means only unigrams, 
    (1, 2) means unigrams and bigrams, and (2, 2) means only bigrams.
    """

    # CountVectorizer converts text into a document-term count matrix
    vectorizer = CountVectorizer(binary=True, stop_words=["english"], ngram_range=(1, 2), lowercase=True, input='filename')
    # fit_transform() learns the vocabulary and returns a CSR sparse matrix
    count_vector = vectorizer.fit_transform(corpus)

    print(f"count_vector sparse matrix in Coordinate Format (COO):\n {count_vector}")

    print(f"\ntype(count_vector) {type(count_vector)}\n")

    print(f"Converting the matrix from sparse to dense:\n")
    dense_array = count_vector.todense()
    print(dense_array)

    # from complementary material -> https://machinelearningmastery.com/sparse-matrices-for-machine-learning/
    # sparsity = 1.0 - count_nonzero(A) / A.size
    sparsity = 1.0 - np.count_nonzero(dense_array) / dense_array.size

    print(f"\nsparsity: {sparsity}, {np.count_nonzero(dense_array)} out of {dense_array.size}\n")

    # this prompts the user?
    print("Enter your search query (use quotation marks to search for multi-word terms, e.g. \"tony blair\" \"george bush\"):")
    print(f"terms in the search query: {terms}")

    # I'll have to check they are len(2) max


    print("initiating boolean_search_result array (set all True, i.e. matching all docs when search query is empty)...")
    boolean_search_result = np.ones((count_vector.shape[0], 1), dtype=bool)
    print(boolean_search_result) 
    
    print(f"boolean_search_result.shape: {boolean_search_result.shape}")

    # about vocabulary:
    """
    vocabularyMapping or iterable, default=None
    Either a Mapping (e.g., a dict) where keys are terms and values are indices in the feature matrix, 
    or an iterable over terms. If not given, a vocabulary is determined from the input documents. 
    Indices in the mapping should not be repeated and should not have any gap between 0 and the largest index.
    """

    # loop throught the terms
    for term in terms:
        print(f"index of \"{term}\" in the dictionary: {vectorizer.vocabulary_.get(term)}")
        print(f"boolean_search_result for \"{term}\":")
        term_column = dense_array[:, vectorizer.vocabulary_.get(term)]
        
        boolean_search_result = np.logical_and(boolean_search_result, term_column.astype(dtype=bool))
        print(boolean_search_result)
        
        print(f"boolean_search_result.shape: {boolean_search_result.shape}")
        print(f"#Matching docs: {np.count_nonzero(boolean_search_result)}")
        #boolean_matrix = np.logical_and(dense_array, dense_array)
        print("\n")



my_search_function(files_list, ['tony blair', 'george bush'])
```
