```python
# https://www.tutorialspoint.com/python/os_walk.htm:
# The Python walk() method of OS module displays the file names in the specified directory tree
#  by traversing the tree either in top-down or bottom-up approach.

import os

files_list = []

for root, dirs, files in os.walk("bbc", topdown=False):
   for name in files:
      #print(os.path.join(root, name))
      files_list.append(os.path.join(root, name))
   #for name in dirs:
   #   print(os.path.join(root, name))

# removing the last document 'bbc/README.TXT'
files_list = files_list[:-1]

print(f"files_list: {files_list}")
print(f"Number of files in the corpus: {len(files_list)}")
```

```python
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer

# about N-grams
"""
ngram_rangetuple (min_n, max_n), default=(1, 1)
The lower and upper boundary of the range of n-values for different word n-grams or char n-grams to be extracted. 
All values of n such such that min_n <= n <= max_n will be used. 
For example an ngram_range of (1, 1) means only unigrams, 
(1, 2) means unigrams and bigrams, and (2, 2) means only bigrams.
"""
# CountVectorizer converts text into a document-term count matrix
vectorizer = CountVectorizer(binary=True, stop_words=["english"], ngram_range=(1, 2))
# fit_transform() learns the vocabulary and returns a CSR sparse matrix
count_vector = vectorizer.fit_transform(files_list)

print(f"count_vector sparse matrix in Coordinate Format (COO):\n {count_vector}")
```

```python
print(f"\ntype(count_vector) {type(count_vector)}\n")

print(f"Converting the matrix from sparse to dense:\n")
dense_array = count_vector.todense()
print(dense_array)

# from complementary material -> https://machinelearningmastery.com/sparse-matrices-for-machine-learning/
# sparsity = 1.0 - count_nonzero(A) / A.size
from numpy import count_nonzero

sparsity = 1.0 - count_nonzero(dense_array) / dense_array.size

print(f"\nsparsity: {sparsity}, {count_nonzero(dense_array)} out of {dense_array.size}")
```
